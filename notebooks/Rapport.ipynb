## Introduction

Le calcul numérique d'intégrales de grande dimension intervient dans de nombreux domaines scientifiques tels que la physique statistique, la finance quantitative, l'apprentissage automatique ou encore la théorie des probabilités.

Lorsque la dimension augmente, les méthodes déterministes classiques de quadrature deviennent rapidement impraticables en raison de la croissance exponentielle du nombre de points nécessaires à l'approximation de l'intégrale. Ce phénomène est connu sous le nom de *malédiction de la dimension*.

Les méthodes de Monte-Carlo constituent une alternative particulièrement efficace dans ce contexte. Leur principe consiste à interpréter l'intégrale recherchée comme une espérance puis à l'estimer à l'aide d'une moyenne empirique construite à partir d'échantillons aléatoires. L'un des principaux avantages de cette approche est que la vitesse de convergence dépend essentiellement du nombre de simulations et reste indépendante de la dimension du problème.

Cependant, cette méthode présente également une limite importante : sa précision est gouvernée par le théorème central limite et l'erreur décroit seulement en \(O(n^{-1/2})\), où \(n\) désigne le nombre d'échantillons simulés. Il est donc souvent nécessaire de mettre en œuvre des techniques de réduction de variance afin d'améliorer l'efficacité des estimateurs obtenus.

L'objectif de ce projet est d'étudier différentes méthodes d'estimation d'intégrales multidimensionnelles. Dans un premier temps, nous mettons en place un estimateur Monte-Carlo standard et analysons sa convergence. Nous introduisons ensuite une méthode de réduction de variance fondée sur l'utilisation de variables de contrôle. Enfin, nous étudions le comportement asymptotique de certaines intégrales lorsque la dimension tend vers l'infini et exploitons cette analyse pour construire des estimateurs plus rapides et plus précis.

L'ensemble des résultats théoriques est accompagné d'implémentations numériques en Python permettant de comparer les performances des différentes approches en termes de précision et de coût de calcul.

## 1. Calcul de l’intégrale par sommes de Riemann.

**T1.** Soit $f$ Lipschitzienne

On considère la grille régulière de $[0,1]^d$
$$
\mathcal G_n = \left\{\left(\frac{a_1}{n},\dots,\frac{a_d}{n}\right)\; ; \;a_1,\dots,a_d \in \{0,\dots,n-1\}\right\}.
$$

Pour
$$
k=\sum_{i=1}^d a_i n^{i-1},
$$
on associe le point
$$
x_k
=
\left(
\frac{a_1}{n},
\dots,
\frac{a_d}{n}
\right).
$$

On définit alors la somme de Riemann
$$
S_n
=
\frac{1}{n^d}
\sum_{k=0}^{n^d-1}
f(x_k).
$$

Pour chaque multi-indice $(a_1,\dots,a_d)$, on considère le cube
$$
Q_k
=
\prod_{i=1}^d
\left[
\frac{a_i}{n},
\frac{a_i+1}{n}
\right].
$$

Les cubes $(Q_k)$ forment une partition de $[0,1]^d$, donc
$$
I(f)
=
\int_{[0,1]^d} f(x)\,dx
=
\sum_{k=0}^{n^d-1}
\int_{Q_k} f(x)\,dx.
$$

Comme $f$ est $K$-Lipschitzienne,
$$
|f(x)-f(y)|
\leq
K\|x-y\|,
\qquad
x,y\in[0,1]^d.
$$

Ainsi, pour tout $x\in Q_k$,
$$
|f(x)-f(x_k)|
\leq
K\|x-x_k\|.
$$

Or le diamètre d’un cube de côté $1/n$ vaut
$$
\frac{\sqrt d}{n},
$$
donc
$$
|f(x)-f(x_k)|
\leq
\frac{K\sqrt d}{n}.
$$

On obtient alors
$$
\begin{aligned}
\left|
\int_{Q_k} f(x)\,dx
-
\frac1{n^d}f(x_k)
\right|
&=
\left|
\int_{Q_k}
\bigl(f(x)-f(x_k)\bigr)\,dx
\right|
\\
&\leq
\int_{Q_k}
|f(x)-f(x_k)|\,dx
\\
&\leq
\frac{K\sqrt d}{n}
\operatorname{Vol}(Q_k).
\end{aligned}
$$

Comme
$$
\operatorname{Vol}(Q_k)=\frac1{n^d},
$$
on a
$$
\left|
\int_{Q_k} f(x)\,dx
-
\frac1{n^d}f(x_k)
\right|
\leq
\frac{K\sqrt d}{n^{d+1}}.
$$

En sommant sur tous les cubes :
$$
\begin{aligned}
|I(f)-S_n|
&\leq
\sum_{k=0}^{n^d-1}
\left|
\int_{Q_k} f(x)\,dx
-
\frac1{n^d}f(x_k)
\right|
\\
&\leq
n^d \times \frac{K\sqrt d}{n^{d+1}}
\\
&=
\frac{K\sqrt d}{n}.
\end{aligned}
$$

Donc
$$
|I(f)-S_n|
\leq
\frac{K\sqrt d}{n},
$$
ce qui montre que
$$
S_n \xrightarrow[n\to\infty]{} I(f),
$$
et même
$$
S_n-I(f)=O(n^{-1}).
$$

**T2.** La somme de Riemann définie par
$$
S_n
=
\frac1{n^d}
\sum_{k=0}^{n^d-1}
f(x_k)
$$
contient exactement
$$
n^d
$$
termes.

Ainsi, le calcul de $S_n$ nécessite l’évaluation de la fonction $f$ en $n^d$ points.

D’après la question précédente,
$$
|S_n-I(f)|
=
O(n^{-1}).
$$

Donc, pour obtenir une précision \(\varepsilon>0\), il faut choisir \(n\) tel que
$$
\frac{C}{n}
\le
\varepsilon,
$$
où \(C>0\) est une constante indépendante de \(n\).

Cela impose
$$
n
=
O(\varepsilon^{-1}).
$$

Or le nombre total d’évaluations est
$$
n^d.
$$

Donc le coût de calcul est
$$
n^d
=
O\left((\varepsilon^{-1})^d\right).
$$

Ainsi, pour estimer \(I(f)\) avec une précision \(\varepsilon\), un temps de calcul en
$$
O(\varepsilon^{-d})
$$
est nécessaire.

C'est la malediction de la dimension.

**S1.**
On considère la fonction
$$
f_1(x)
=
\frac1{
1+\frac1d\sum_{i=1}^d x_i^4
+
\left(
\frac1d\sum_{i=1}^d x_i
\right)^4
}.
$$

On souhaite approximer
$$
I(f_1)
=
\int_{[0,1]^d} f_1(x)\,dx
$$
par une somme de Riemann.

Voici un code Python utilisant une grille régulière.


```python
import numpy as np
import itertools
import time

def f1(x):
    d = len(x)
    return 1 / (
        1
        + np.mean(x**4)
        + (np.mean(x))**4
    )

def riemann_sum(n, d):
    grid = np.linspace(0, 1, n, endpoint=False)

    total = 0.0

    for point in itertools.product(grid, repeat=d):
        x = np.array(point)
        total += f1(x)

    return total / (n**d)

n = 100

for d in range(1, 8):

    start = time.time()

    result = riemann_sum(n, d)

    elapsed = time.time() - start

    print(
        f"d = {d}, "
        f"I ≈ {result:.6f}, "
        f"temps = {elapsed:.2f} s"
    )
```

**S2.**
On peut utiliser une méthode d’intégration numérique adaptative de Python, par exemple `nquad` de `scipy.integrate`.

Voici un exemple de code :


```python
import numpy as np
import time
from scipy.integrate import nquad

def f1(*x):

    x = np.array(x)

    d = len(x)

    return 1 / (
        1
        + np.mean(x**4)
        + (np.mean(x))**4
    )

for d in range(1, 8):

    bounds = [(0, 1)] * d

    start = time.time()

    result, error = nquad(f1, bounds)

    elapsed = time.time() - start

    print(
        f"d = {d}, "
        f"I ≈ {result:.6f}, "
        f"erreur ≈ {error:.2e}, "
        f"temps = {elapsed:.2f} s"
    )
```

## 2. Calcul de l’intégrale par méthode de Monte-Carlo.

**T3.**
On considère une suite $(X_i)_{i\ge1}$ de variables aléatoires i.i.d. de loi uniforme sur $[0,1]^d$.

On pose
$$
M_n(f)
=
\frac1n
\sum_{k=1}^n f(X_k).
$$

Comme les $X_k$ sont uniformes sur $[0,1]^d$,
$$
\mathbb E[f(X_k)]
=
\int_{[0,1]^d} f(x)\,dx
=
I(f).
$$

Les variables $f(X_k)$ sont i.i.d. et intégrables puisque
$$
\int_{[0,1]^d} f^2(x)\,dx < +\infty.
$$

En particulier,
$$
\mathbb E[|f(X_k)|] < +\infty.
$$

Par la loi forte des grands nombres,
$$
M_n(f)
\xrightarrow[n\to\infty]{p.s.}
I(f).
$$

Autrement dit,
$$
M_n(f)
\to
I(f)
\quad \text{presque sûrement}.
$$

Ainsi, la méthode de Monte-Carlo fournit un estimateur convergent de $I(f)$.

**T4.**
On suppose
$$
\int_{[0,1]^d} f^2(x)\,dx
<
+\infty.
$$

On pose
$$
\sigma_f^2
=
\int_{[0,1]^d} f^2(x)\,dx
-
I(f)^2.
$$

Comme les variables $f(X_k)$ sont i.i.d. de variance $\sigma_f^2$, le théorème central limite donne
$$
\sqrt n\bigl(M_n(f)-I(f)\bigr)
\xrightarrow[n\to\infty]{\mathcal L}
\mathcal N(0,\sigma_f^2).
$$

Ainsi, pour \(n\) grand,
$$
M_n(f)-I(f)
\approx
\mathcal N
\left(
0,
\frac{\sigma_f^2}{n}
\right).
$$

On obtient alors un intervalle de confiance asymptotique à \(95\%\) :
$$
I(f)
\in
\left[
M_n(f)
-
1.96\frac{\sigma_f}{\sqrt n},
\;
M_n(f)
+
1.96\frac{\sigma_f}{\sqrt n}
\right].
$$

En pratique, \(\sigma_f\) est inconnue et on utilise l’estimateur empirique
$$
\hat\sigma_n^2
=
\frac1{n-1}
\sum_{k=1}^n
\bigl(f(X_k)-M_n(f)\bigr)^2.
$$

L’intervalle de confiance devient alors
$$
\left[
M_n(f)
-
1.96\frac{\hat\sigma_n}{\sqrt n},
\;
M_n(f)
+
1.96\frac{\hat\sigma_n}{\sqrt n}
\right].
$$

**T5.**
 L’intervalle de confiance obtenu précédemment a une largeur proportionnelle à
$$
\frac{\sigma_f}{\sqrt n}.
$$

Ainsi, pour obtenir une précision \(\varepsilon>0\), il faut imposer
$$
\frac1{\sqrt n}
=
O(\varepsilon).
$$

Cela donne
$$
n
=
O(\varepsilon^{-2}).
$$

Comme le coût de calcul de la méthode de Monte-Carlo est proportionnel au nombre d’échantillons \(n\), le temps de calcul nécessaire est
$$
O(\varepsilon^{-2}).
$$

Point crucial : ce coût ne dépend pas de la dimension \(d\).

Au contraire, pour les sommes de Riemann, on avait obtenu
$$
O(\varepsilon^{-d}).
$$

Ainsi :

- les méthodes déterministes deviennent rapidement impossibles lorsque \(d\) augmente ;
- les méthodes de Monte-Carlo gardent un coût polynomial indépendant de la dimension.

C’est la raison fondamentale pour laquelle Monte-Carlo est utilisé en grande dimension.

**S3.**
Le théorème central limite donne
$$
\sqrt n
\frac{
M_n(f)-I(f)
}{
\sigma_f
}
\xrightarrow[n\to\infty]{\mathcal L}
\mathcal N(0,1).
$$

Comme
$$
\widehat\sigma_n^2
=
M_n(f^2)-M_n(f)^2
\xrightarrow[n\to\infty]{a.s.}
\sigma_f^2,
$$
le théorème de Slutsky permet d’obtenir
$$
\sqrt{
\frac{n}{
M_n(f^2)-M_n(f)^2
}
}
\bigl(M_n(f)-I(f)\bigr)
\xrightarrow[n\to\infty]{\mathcal L}
\mathcal N(0,1).
$$

On peut alors construire un intervalle de confiance asymptotique à \(95\%\) :
$$
\left[
M_n(f)
-
1.96
\sqrt{
\frac{
M_n(f^2)-M_n(f)^2
}{n}
},
\;
M_n(f)
+
1.96
\sqrt{
\frac{
M_n(f^2)-M_n(f)^2
}{n}
}
\right].
$$

```python
import numpy as np
import time

def f1(x):

    d = len(x)

    return 1 / (
        1
        + np.mean(x**4)
        + (np.mean(x))**4
    )

def monte_carlo(n, d):

    start = time.time()

    X = np.random.rand(n, d)

    values = np.array([f1(x) for x in X])

    Mn = np.mean(values)

    Mn2 = np.mean(values**2)

    sigma = np.sqrt(Mn2 - Mn**2)

    radius = 1.96 * sigma / np.sqrt(n)

    elapsed = time.time() - start

    return Mn, radius, elapsed

n = 10000

dimensions = [1, 2, 3, 10, 50, 100, 1000]

for d in dimensions:

    estimate, radius, elapsed = monte_carlo(n, d)

    print(
        f"d = {d:4d} | "
        f"I ≈ {estimate:.6f} | "
        f"IC95 = [{estimate-radius:.6f}, {estimate+radius:.6f}] | "
        f"temps = {elapsed:.4f} s"
    )
```

***Analyse des résultats***

Contrairement aux sommes de Riemann, le coût de Monte-Carlo ne dépend pas exponentiellement de la dimension.

Le nombre d’évaluations reste fixé à
$$
n=10000,
$$
quelle que soit la valeur de \(d\).

Ainsi :

- les dimensions \(d=50\), \(100\) ou \(1000\) restent parfaitement calculables ;
- le temps de calcul augmente peu avec \(d\) ;
- la précision reste gouvernée par
$$
O(n^{-1/2}),
$$
indépendamment de la dimension.

C’est l’avantage majeur des méthodes de Monte-Carlo.

## 3. Méthode de réduction de variance

**T6.**
On considère
$$
f_2(x)
=
\frac1d\sum_{i=1}^d x_i,
$$
où $X=(X_1,\ldots,X_d)$ suit la loi uniforme sur $[0,1]^d$.

Comme les coordonnées sont indépendantes et uniformes sur $[0,1]$,
$$
\mathbb E[X_i]=\frac12.
$$

Par linéarité de l'espérance,
$$
I(f_2)
=
\mathbb E[f_2(X)]
=
\frac1d\sum_{i=1}^d \mathbb E[X_i]
=
\frac1d \times d \times \frac12
=
\frac12.
$$

De plus,
$$
\operatorname{Var}(X_i)
=
\frac1{12}.
$$

Par indépendance,
$$
\operatorname{Var}(f_2(X))
=
\operatorname{Var}
\left(
\frac1d\sum_{i=1}^d X_i
\right)
=
\frac1{d^2}
\sum_{i=1}^d
\operatorname{Var}(X_i).
$$

Ainsi,
$$
\operatorname{Var}(f_2(X))
=
\frac1{d^2}
\times d
\times \frac1{12}
=
\frac1{12d}.
$$

On obtient finalement
$$
I(f_2)=\frac12,
\qquad
\operatorname{Var}(f_2(X))
=
\frac1{12d}.
$$

**T7.**
Posons
$$
Y_\lambda
=
f_1(X)+\lambda f_2(X).
$$

On cherche à minimiser
$$
V(\lambda)
=
\operatorname{Var}(Y_\lambda).
$$

Par bilinéarité de la covariance,
$$
V(\lambda)
=
\operatorname{Var}(f_1)
+
2\lambda
\operatorname{Cov}(f_1,f_2)
+
\lambda^2
\operatorname{Var}(f_2).
$$

Comme \(V\) est un polynôme du second degré convexe, son minimum est atteint lorsque
$$
V'(\lambda)=0.
$$

Or
$$
V'(\lambda)
=
2\operatorname{Cov}(f_1,f_2)
+
2\lambda
\operatorname{Var}(f_2).
$$

Ainsi
$$
\lambda^\ast
=
-
\frac{
\operatorname{Cov}(f_1,f_2)
}{
\operatorname{Var}(f_2)
}.
$$

D'après la question précédente,
$$
\operatorname{Var}(f_2)
=
\frac1{12d}.
$$

Donc
$$
\lambda^\ast
=
-12d\,
\operatorname{Cov}(f_1,f_2).
$$

Or
$$
\operatorname{Cov}(f_1,f_2)
=
\mathbb E[f_1f_2]
-
\mathbb E[f_1]\mathbb E[f_2].
$$

On peut l'estimer par
$$
M_n(f_1f_2)-M_n(f_1)M_n(f_2).
$$

On définit alors
$$
\widehat\lambda_n
=
12d
\Bigl(
M_n(f_1)M_n(f_2)
-
M_n(f_1f_2)
\Bigr).
$$

Par la loi forte des grands nombres,
$$
M_n(f_1)
\to
\mathbb E[f_1],
$$

$$
M_n(f_2)
\to
\mathbb E[f_2],
$$

et
$$
M_n(f_1f_2)
\to
\mathbb E[f_1f_2]
\qquad
\text{p.s.}
$$

Ainsi
$$
\widehat\lambda_n
\xrightarrow[n\to\infty]{a.s.}
12d
\Bigl(
\mathbb E[f_1]\mathbb E[f_2]
-
\mathbb E[f_1f_2]
\Bigr).
$$

Or
$$
\lambda^\ast
=
-12d
\Bigl(
\mathbb E[f_1f_2]
-
\mathbb E[f_1]\mathbb E[f_2]
\Bigr).
$$

Donc
$$
\widehat\lambda_n
\xrightarrow[n\to\infty]{p.s.}
\lambda^\ast.
$$

**T8.**

On introduit l'estimateur à variable de contrôle

$$
\widetilde M_n
=
M_n(f_1)
+
\widehat\lambda_n
\left(
M_n(f_2)-\frac12
\right).
$$

Comme
$$
\mathbb E[f_2]
=
\frac12,
$$

on a

$$
\mathbb E[\widetilde M_n]
=
\mathbb E[M_n(f_1)]
=
I(f_1).
$$

L'estimateur reste donc sans biais.

Par la question précédente,
$$
\widehat\lambda_n
\xrightarrow[]{a.s.}
\lambda^\ast.
$$

Le théorème central limite appliqué à

$$
f_1+\lambda^\ast f_2
$$

donne

$$
\sqrt n
\left(
\widetilde M_n-I(f_1)
\right)
\xrightarrow[]{\mathcal L}
\mathcal N(0,\widetilde\sigma^2),
$$

où

$$
\widetilde\sigma^2
=
\operatorname{Var}
\bigl(
f_1+\lambda^\ast f_2
\bigr).
$$

Or

$$
\widetilde\sigma^2
=
\operatorname{Var}(f_1)
-
\frac{
\operatorname{Cov}(f_1,f_2)^2
}{
\operatorname{Var}(f_2)
}.
$$

Comme

$$
\operatorname{Var}(f_2)
=
\frac1{12d},
$$

on obtient

$$
\widetilde\sigma^2
=
\operatorname{Var}(f_1)
-
12d\,
\operatorname{Cov}(f_1,f_2)^2.
$$

L'estimateur empirique proposé est

$$
\widetilde V_n
=
M_n(f_1^2)
-
M_n(f_1)^2
-
\frac{\widehat\lambda_n^2}{12d}.
$$

Par la loi forte des grands nombres,

$$
\widetilde V_n
\xrightarrow[]{a.s.}
\widetilde\sigma^2.
$$

Le théorème de Slutsky permet alors d'obtenir

$$
\sqrt{
\frac n{\widetilde V_n}
}
\Bigl(
\widetilde M_n-I(f_1)
\Bigr)
\xrightarrow[]{\mathcal L}
\mathcal N(0,1).
$$

On en déduit l'intervalle de confiance asymptotique à \(95\%\)

$$
\left[
\widetilde M_n
-
1.96
\sqrt{\frac{\widetilde V_n}{n}},
\;
\widetilde M_n
+
1.96
\sqrt{\frac{\widetilde V_n}{n}}
\right].
$$

**S4.**
L'objectif est d'améliorer la précision de l'estimateur Monte-Carlo classique

$$
M_n(f_1)
=
\frac1n
\sum_{k=1}^{n}
f_1(X_k),
$$

en utilisant une variable de contrôle.

On considère la fonction

$$
f_2(x)
=
\frac1d
\sum_{i=1}^{d}x_i,
$$

dont l'intégrale est connue :

$$
I(f_2)=\frac12.
$$

D'après la question T7, le coefficient optimal est estimé par

$$
\widehat\lambda_n
=
12d
\Bigl(
M_n(f_1)M_n(f_2)
-
M_n(f_1f_2)
\Bigr).
$$

On construit alors l'estimateur

$$
\widetilde M_n
=
M_n(f_1)
+
\widehat\lambda_n
\left(
M_n(f_2)-\frac12
\right).
$$

D'après la question T8, un estimateur de la variance asymptotique est

$$
\widetilde V_n
=
M_n(f_1^2)
-
M_n(f_1)^2
-
\frac{\widehat\lambda_n^2}{12d}.
$$

L'intervalle de confiance asymptotique à 95 % est donc

$$
\left[
\widetilde M_n
-
1.96
\sqrt{\frac{\widetilde V_n}{n}},
\;
\widetilde M_n
+
1.96
\sqrt{\frac{\widetilde V_n}{n}}
\right].
$$

Nous appliquons cette méthode pour

$$
n=10000
$$

et

$$
d\in\{1,2,3,10,50,100,1000\}.
$$

```python
import numpy as np
import pandas as pd
import time

def f1(X):

    m1 = np.mean(X, axis=1)
    m4 = np.mean(X**4, axis=1)

    return 1/(1 + m4 + m1**4)

def f2(X):

    return np.mean(X, axis=1)

def control_variate(n,d):

    start = time.time()

    X = np.random.rand(n,d)

    F1 = f1(X)
    F2 = f2(X)

    Mn_f1 = np.mean(F1)
    Mn_f2 = np.mean(F2)

    Mn_f1_sq = np.mean(F1**2)
    Mn_f1f2 = np.mean(F1*F2)

    lambda_hat = 12*d*(Mn_f1*Mn_f2 - Mn_f1f2)

    M_tilde = Mn_f1 + lambda_hat*(Mn_f2 - 0.5)

    V_tilde = (
        Mn_f1_sq
        - Mn_f1**2
        - lambda_hat**2/(12*d)
    )

    radius = 1.96*np.sqrt(V_tilde/n)

    elapsed = time.time() - start

    return [
        d,
        M_tilde,
        M_tilde-radius,
        M_tilde+radius,
        2*radius,
        elapsed
    ]

n = 10000

results = []

for d in [1,2,3,10,50,100,1000]:
    results.append(control_variate(n,d))

df = pd.DataFrame(
    results,
    columns=[
        "d",
        "Estimateur",
        "Borne inf",
        "Borne sup",
        "Largeur IC",
        "Temps (s)"
    ]
)

df
```

#### Analyse

On constate que la méthode à variable de contrôle produit des intervalles de confiance plus étroits que la méthode Monte-Carlo classique.

Cette amélioration provient de la forte corrélation entre \(f_1\) et \(f_2\), qui permet de réduire la variance de l'estimateur.

Le coût supplémentaire est négligeable puisque les quantités

$$
M_n(f_2)
\qquad\text{et}\qquad
M_n(f_1f_2)
$$

sont calculées à partir des mêmes échantillons.

La précision est donc améliorée sans augmentation significative du temps de calcul.

**S5.** Approximation de $\lambda^\ast$

Dans la méthode précédente, le coefficient

$$
\widehat\lambda_n
=
12d
\Bigl(
M_n(f_1)M_n(f_2)
-
M_n(f_1f_2)
\Bigr)
$$

est estimé à partir des \(n\) échantillons.

On propose ici de l'approximer une seule fois à l'aide d'un échantillon de taille 100.

On note cette approximation

$$
\widehat\lambda_{100}.
$$

L'estimateur devient alors

$$
\widetilde M'_n
=
M_n(f_1)
+
\widehat\lambda_{100}
\left(
M_n(f_2)-\frac12
\right).
$$

L'estimateur de variance associé est

$$
\widetilde V'_n
=
M_n(f_1^2)
-
M_n(f_1)^2
+
2\widehat\lambda_{100}
\Bigl(
M_n(f_1f_2)
-
M_n(f_1)M_n(f_2)
\Bigr)
+
\frac{\widehat\lambda_{100}^2}{12d}.
$$

L'intervalle de confiance à 95 % est

$$
\left[
\widetilde M'_n
-
1.96
\sqrt{\frac{\widetilde V'_n}{n}},
\;
\widetilde M'_n
+
1.96
\sqrt{\frac{\widetilde V'_n}{n}}
\right].
$$

L'objectif est de comparer cette méthode avec celle de la question précédente.

```python
import numpy as np
import pandas as pd
import time

def control_variate_fast(n,d):

    X0 = np.random.rand(100,d)

    F10 = f1(X0)
    F20 = f2(X0)

    lambda100 = 12*d*(
        np.mean(F10)*np.mean(F20)
        -
        np.mean(F10*F20)
    )

    start = time.time()

    X = np.random.rand(n,d)

    F1 = f1(X)
    F2 = f2(X)

    Mn_f1 = np.mean(F1)
    Mn_f2 = np.mean(F2)

    Mn_f1_sq = np.mean(F1**2)
    Mn_f1f2 = np.mean(F1*F2)

    Mprime = Mn_f1 + lambda100*(Mn_f2 - 0.5)

    Vprime = (
        Mn_f1_sq
        - Mn_f1**2
        + 2*lambda100*(Mn_f1f2 - Mn_f1*Mn_f2)
        + lambda100**2/(12*d)
    )

    radius = 1.96*np.sqrt(Vprime/n)

    elapsed = time.time() - start

    return [
        d,
        Mprime,
        Mprime-radius,
        Mprime+radius,
        2*radius,
        elapsed
    ]

results = []

for d in [1,2,3,10,50,100,1000]:
    results.append(control_variate_fast(10000,d))

df = pd.DataFrame(
    results,
    columns=[
        "d",
        "Estimateur",
        "Borne inf",
        "Borne sup",
        "Largeur IC",
        "Temps (s)"
    ]
)

df
```

#### Analyse

Les résultats obtenus sont très proches de ceux de la méthode utilisant $\widehat\lambda_n$.

La raison est que $\widehat\lambda_{100}$ constitue déjà une approximation raisonnable du coefficient optimal $\lambda^\ast$.

Le gain en temps de calcul reste limité car l'estimation de $\widehat\lambda_n$ n'était déjà pas très coûteuse.

En revanche, la qualité de l'intervalle de confiance demeure comparable à celle obtenue avec la méthode précédente.

Cette approche permet donc de conserver l'essentiel du gain de réduction de variance tout en simplifiant légèrement les calculs.

## 4. Calcul approché en dimension $d$ grande

**T9.**
Soit \(U_1,\ldots,U_d\) des variables aléatoires i.i.d. de loi uniforme sur \([0,1]\).

Par indépendance, le vecteur

$$
(U_1,\ldots,U_d)
$$

admet pour densité

$$
f(u_1,\ldots,u_d)
=
\prod_{i=1}^{d} 1_{[0,1]}(u_i)
=
1_{[0,1]^d}(u_1,\ldots,u_d).
$$

Il s'agit précisément de la densité de la loi uniforme sur le cube \([0,1]^d\).

Ainsi,

$$
(U_1,\ldots,U_d)
\overset{\mathcal L}{=}
X_1.
$$

Par conséquent,

$$
I(f_1)
=
\mathbb E[f_1(X_1)].
$$

En remplaçant \(X_1\) par \((U_1,\ldots,U_d)\), on obtient

$$
I(f_1)
=
\mathbb E\!\left[
\frac{1}{
1+\frac1d\sum_{i=1}^{d}U_i^4
+
\left(
\frac1d\sum_{i=1}^{d}U_i
\right)^4
}
\right].
$$

**T10.**
On considère les deux moyennes empiriques

$$
A_d
=
\frac1d
\sum_{i=1}^{d}
U_i
$$

et

$$
B_d
=
\frac1d
\sum_{i=1}^{d}
U_i^4.
$$

Comme les variables \(U_i\) sont i.i.d. intégrables,

la loi forte des grands nombres donne

$$
A_d
\xrightarrow[d\to\infty]{a.s.}
\mathbb E[U_1]
$$

et

$$
B_d
\xrightarrow[d\to\infty]{a.s.}
\mathbb E[U_1^4].
$$

Or

$$
\mathbb E[U_1]
=
\int_0^1 x\,dx
=
\frac12,
$$

et

$$
\mathbb E[U_1^4]
=
\int_0^1 x^4\,dx
=
\frac15.
$$

Ainsi,

$$
\frac1d
\sum_{i=1}^{d}
U_i
\xrightarrow[d\to\infty]{a.s.}
\frac12,
$$

et

$$
\frac1d
\sum_{i=1}^{d}
U_i^4
\xrightarrow[d\to\infty]{a.s.}
\frac15.
$$

Autrement dit,

$$
\left(
\frac1d\sum_{i=1}^{d}U_i,
\,
\frac1d\sum_{i=1}^{d}U_i^4
\right)
\xrightarrow[d\to\infty]{a.s.}
\left(
\frac12,
\frac15
\right).
$$

**T11.**
On pose

$$
Y_i
=
\begin{pmatrix}
U_i\\
U_i^4
\end{pmatrix}.
$$

Les variables \(Y_i\) sont i.i.d. dans \(\mathbb R^2\).

On a

$$
\mathbb E[Y_i]
=
\begin{pmatrix}
\frac12\\
\frac15
\end{pmatrix}.
$$

Le théorème central limite multidimensionnel donne

$$
\sqrt d
\left(
\frac1d\sum_{i=1}^{d}Y_i
-
\mathbb E[Y_i]
\right)
\xrightarrow[d\to\infty]{\mathcal L}
\mathcal N(0,\Sigma),
$$

où \(\Sigma\) est la matrice de covariance de \(Y_1\).

Calculons ses coefficients.

Tout d'abord,

$$
\operatorname{Var}(U_1)
=
\frac1{12}.
$$

Ensuite,

$$
\operatorname{Var}(U_1^4)
=
\mathbb E[U_1^8]
-
\mathbb E[U_1^4]^2.
$$

Or

$$
\mathbb E[U_1^8]
=
\int_0^1 x^8\,dx
=
\frac19.
$$

Ainsi,

$$
\operatorname{Var}(U_1^4)
=
\frac19-\frac1{25}
=
\frac{16}{225}.
$$

Enfin,

$$
\operatorname{Cov}(U_1,U_1^4)
=
\mathbb E[U_1^5]
-
\mathbb E[U_1]\mathbb E[U_1^4].
$$

Comme

$$
\mathbb E[U_1^5]
=
\frac16,
$$

on obtient

$$
\operatorname{Cov}(U_1,U_1^4)
=
\frac16-\frac1{10}
=
\frac1{15}.
$$

La matrice de covariance vaut donc

$$
\Sigma
=
\begin{pmatrix}
\frac1{12} & \frac1{15}\\
\frac1{15} & \frac{16}{225}
\end{pmatrix}.
$$

Par conséquent,

$$
\sqrt d
\begin{pmatrix}
\frac1d\sum_{i=1}^{d}U_i-\frac12\\[2mm]
\frac1d\sum_{i=1}^{d}U_i^4-\frac15
\end{pmatrix}
\xrightarrow[d\to\infty]{\mathcal L}
\mathcal N(0,\Sigma).
$$

Autrement dit,

$$
\sqrt d
\left(
\frac1d\sum_{i=1}^{d}U_i-\frac12,
\,
\frac1d\sum_{i=1}^{d}U_i^4-\frac15
\right)
\xrightarrow[d\to\infty]{\mathcal L}
(G_1,G_2),
$$

où \((G_1,G_2)\) est un vecteur gaussien centré de matrice de covariance

$$
\Sigma
=
\begin{pmatrix}
\frac1{12} & \frac1{15}\\
\frac1{15} & \frac{16}{225}
\end{pmatrix}.
$$

**S6.**

D'après la question T11,

$$
\sqrt d
\left(
\frac1d\sum_{i=1}^{d}U_i-\frac12,
\,
\frac1d\sum_{i=1}^{d}U_i^4-\frac15
\right)
\xrightarrow[d\to\infty]{\mathcal L}
(G_1,G_2),
$$

où \((G_1,G_2)\) est un vecteur gaussien centré de matrice de covariance

$$
\Sigma=
\begin{pmatrix}
\frac1{12} & \frac1{15}\\
\frac1{15} & \frac{16}{225}
\end{pmatrix}.
$$

On obtient alors l'approximation

$$
\widehat I(f_1)
=
\mathbb E\!\left[
\frac1{
1+\frac15+\frac1{\sqrt d}G_2
+
\left(
\frac12+\frac1{\sqrt d}G_1
\right)^4
}
\right].
$$

Cette espérance porte uniquement sur une loi gaussienne bidimensionnelle.

On peut donc l'estimer efficacement par Monte-Carlo en simulant directement des réalisations de \((G_1,G_2)\).

On utilise l'estimateur

$$
\widehat M_n
=
\frac1n
\sum_{k=1}^{n}
h(G^{(k)}_1,G^{(k)}_2),
$$

où

$$
h(g_1,g_2)
=
\frac1{
1+\frac15+\frac1{\sqrt d}g_2
+
\left(
\frac12+\frac1{\sqrt d}g_1
\right)^4
}.
$$

Le théorème central limite permet ensuite de construire un intervalle de confiance à 95%.

```python
import numpy as np
import pandas as pd
import time

Sigma = np.array([
    [1/12, 1/15],
    [1/15, 16/225]
])

def approx_I_hat(d, n=10000):

    start = time.time()

    G = np.random.multivariate_normal(
        mean=[0,0],
        cov=Sigma,
        size=n
    )

    G1 = G[:,0]
    G2 = G[:,1]

    values = 1 / (
        1
        + 1/5
        + G2/np.sqrt(d)
        + (0.5 + G1/np.sqrt(d))**4
    )

    estimate = np.mean(values)

    variance = np.var(values, ddof=1)

    radius = 1.96*np.sqrt(variance/n)

    elapsed = time.time() - start

    return [
        d,
        estimate,
        estimate-radius,
        estimate+radius,
        elapsed
    ]

results = []

for d in [100,1000]:
    results.append(approx_I_hat(d))

pd.DataFrame(
    results,
    columns=[
        "d",
        "Estimateur",
        "Borne inf",
        "Borne sup",
        "Temps (s)"
    ]
)
```

## Analyse

Cette méthode remplace la simulation dans un espace de dimension \(d\) par la simulation d'un vecteur gaussien de dimension 2.

Le coût de calcul devient donc indépendant de \(d\).

Pour \(d=100\) et \(d=1000\), cette approche est significativement plus rapide que les méthodes Monte-Carlo précédentes.

Les résultats obtenus doivent être comparés aux intervalles de confiance calculés aux questions S4 et S5.

On observe généralement que l'approximation devient très précise lorsque \(d\) est grand, conformément aux résultats asymptotiques obtenus aux questions T10 et T11.

Comme l'intégrale est désormais de dimension 2, on peut utiliser directement les routines d'intégration adaptative de SciPy.

```python
import numpy as np
from scipy.integrate import dblquad

Sigma = np.array([
    [1/12,1/15],
    [1/15,16/225]
])

Sigma_inv = np.linalg.inv(Sigma)
Sigma_det = np.linalg.det(Sigma)

def gaussian_density(g2,g1):

    x = np.array([g1,g2])

    return (
        np.exp(
            -0.5*x@Sigma_inv@x
        )
        /
        (2*np.pi*np.sqrt(Sigma_det))
    )

def integrand(g2,g1,d):

    h = 1 / (
        1
        + 1/5
        + g2/np.sqrt(d)
        + (0.5 + g1/np.sqrt(d))**4
    )

    return h*gaussian_density(g2,g1)

for d in [100,1000]:

    value, error = dblquad(
        lambda g2,g1: integrand(g2,g1,d),
        -8,
        8,
        lambda x:-8,
        lambda x:8
    )

    print(
        f"d={d}, "
        f"Ihat={value:.8f}, "
        f"error={error:.2e}"
    )
```

L'objectif est de vérifier que l'approximation asymptotique obtenue lorsque \(d\to\infty\) fournit une estimation fidèle de l'intégrale initiale tout en réduisant considérablement le coût de calcul.

Nous implémentons maintenant cette méthode à l'aide des fonctions d'intégration numérique de la bibliothèque SciPy.

```python
import numpy as np
import pandas as pd
import time
from scipy.integrate import dblquad

# Matrice de covariance obtenue en T11

Sigma = np.array([
    [1/12, 1/15],
    [1/15, 16/225]
])

Sigma_inv = np.linalg.inv(Sigma)
Sigma_det = np.linalg.det(Sigma)

# Densité gaussienne bidimensionnelle

def gaussian_density(g1, g2):

    x = np.array([g1, g2])

    exponent = -0.5 * x @ Sigma_inv @ x

    return (
        np.exp(exponent)
        /
        (2 * np.pi * np.sqrt(Sigma_det))
    )

# Fonction à intégrer

def integrand(g2, g1, d):

    value = 1 / (
        1
        + 1/5
        + g2/np.sqrt(d)
        + (0.5 + g1/np.sqrt(d))**4
    )

    return value * gaussian_density(g1, g2)

# Approximation numérique de I_hat(f1)

def compute_I_hat(d):

    start = time.time()

    value, error = dblquad(
        lambda g2, g1: integrand(g2, g1, d),
        -8,
        8,
        lambda g1: -8,
        lambda g1: 8
    )

    elapsed = time.time() - start

    return [
        d,
        value,
        error,
        elapsed
    ]

# Calcul pour les dimensions demandées

results = []

for d in [100, 1000]:

    results.append(compute_I_hat(d))

df = pd.DataFrame(
    results,
    columns=[
        "d",
        "Approximation",
        "Erreur estimée",
        "Temps (s)"
    ]
)

df
```

## Analyse des résultats

L'intégration numérique est réalisée sur un problème de dimension 2, indépendamment de la dimension initiale \(d\).

Contrairement aux méthodes Monte-Carlo étudiées précédemment, le coût de calcul ne dépend donc plus de la dimension de l'intégrale de départ.

Les valeurs obtenues pour \(d=100\) et \(d=1000\) sont très proches de celles fournies par les estimateurs Monte-Carlo. Cette observation confirme la validité de l'approximation asymptotique dérivée à partir du théorème central limite.

On constate également que la précision de cette approximation s'améliore lorsque la dimension augmente. En effet, les termes négligés dans le développement asymptotique sont d'ordre inférieur à \(1/d\) et deviennent rapidement insignifiants.

Cette approche permet ainsi de remplacer un problème d'intégration de grande dimension par le calcul d'une intégrale bidimensionnelle beaucoup plus simple à traiter numériquement.

**S8.**
On considère la fonction

$$
f_3(x)
=
\frac{
\left(\sum_{i=1}^{d} x_i^2\right)^2
}{
1+
\left(\sum_{i=1}^{d} x_i^3\right)
\left(\sum_{i=1}^{d} x_i^4\right)
}.
$$

L'objectif est de construire un estimateur à réduction de variance à l'aide d'une variable de contrôle \(f_4\) dont l'intégrale est connue explicitement.

Nous choisissons

$$
f_4(x)
=
\left(
\sum_{i=1}^{d}x_i^2
\right)^2.
$$

Ce choix est naturel puisque \(f_4\) correspond exactement au numérateur de \(f_3\). On s'attend donc à une forte corrélation entre \(f_3\) et \(f_4\), ce qui devrait permettre une réduction significative de la variance.

Calculons l'intégrale de \(f_4\).

En développant le carré,

$$
f_4(x)
=
\sum_{i=1}^{d}x_i^4
+
2\sum_{1\le i<j\le d}x_i^2x_j^2.
$$

Par linéarité de l'intégrale,

$$
I(f_4)
=
d\,\mathbb E[U^4]
+
2\binom d2 \mathbb E[U^2]^2,
$$

où \(U\sim\mathcal U([0,1])\).

Or

$$
\mathbb E[U^2]
=
\int_0^1 x^2\,dx
=
\frac13,
$$

et

$$
\mathbb E[U^4]
=
\int_0^1 x^4\,dx
=
\frac15.
$$

On obtient donc

$$
I(f_4)
=
\frac d5
+
\frac{d(d-1)}9.
$$

Ainsi,

$$
\boxed{
I(f_4)
=
\frac d5+\frac{d(d-1)}9
}.
$$

Nous utilisons alors l'estimateur à variable de contrôle

$$
\widetilde M_n
=
M_n(f_3)
+
\widehat\lambda_n
\bigl(
M_n(f_4)-I(f_4)
\bigr),
$$

où

$$
\widehat\lambda_n
=
\frac{
M_n(f_3)M_n(f_4)-M_n(f_3f_4)
}{
M_n(f_4^2)-M_n(f_4)^2
}.
$$

Nous comparerons ensuite les intervalles de confiance obtenus avec cet estimateur à ceux produits par la méthode Monte-Carlo standard.

```python
import numpy as np
import pandas as pd
import time

# ============================================================
# Fonction f3
# ============================================================

def f3(X):

    S2 = np.sum(X**2, axis=1)
    S3 = np.sum(X**3, axis=1)
    S4 = np.sum(X**4, axis=1)

    return S2**2 / (1 + S3*S4)

# ============================================================
# Variable de contrôle f4
# ============================================================

def f4(X):

    S2 = np.sum(X**2, axis=1)

    return S2**2

# ============================================================
# Valeur exacte de I(f4)
# ============================================================

def I_f4(d):

    return d/5 + d*(d-1)/9

# ============================================================
# Monte-Carlo standard
# ============================================================

def monte_carlo_standard(n, d):

    start = time.time()

    X = np.random.rand(n, d)

    F3 = f3(X)

    estimate = np.mean(F3)

    variance = np.var(F3, ddof=1)

    radius = 1.96 * np.sqrt(variance / n)

    elapsed = time.time() - start

    return {
        "estimate": estimate,
        "radius": radius,
        "time": elapsed
    }

# ============================================================
# Monte-Carlo avec variable de contrôle
# ============================================================

def monte_carlo_control(n, d):

    start = time.time()

    X = np.random.rand(n, d)

    F3 = f3(X)
    F4 = f4(X)

    Mn_f3 = np.mean(F3)
    Mn_f4 = np.mean(F4)

    Mn_f3f4 = np.mean(F3 * F4)

    Var_f4 = np.var(F4, ddof=1)

    # estimateur du coefficient optimal

    lambda_hat = -(
        Mn_f3f4 - Mn_f3 * Mn_f4
    ) / Var_f4

    # estimateur corrigé

    estimate = (
        Mn_f3
        + lambda_hat * (Mn_f4 - I_f4(d))
    )

    # variance empirique du nouvel estimateur

    Z = F3 + lambda_hat * (F4 - I_f4(d))

    variance = np.var(Z, ddof=1)

    radius = 1.96 * np.sqrt(variance / n)

    elapsed = time.time() - start

    return {
        "estimate": estimate,
        "radius": radius,
        "time": elapsed
    }

# ============================================================
# Comparaison
# ============================================================

n = 10000

results = []

for d in [1, 2, 3, 10, 50, 100]:

    mc = monte_carlo_standard(n, d)

    cv = monte_carlo_control(n, d)

    gain = mc["radius"] / cv["radius"]

    results.append([
        d,
        mc["estimate"],
        2 * mc["radius"],
        cv["estimate"],
        2 * cv["radius"],
        gain,
        cv["time"]
    ])

df = pd.DataFrame(
    results,
    columns=[
        "d",
        "MC standard",
        "Largeur IC MC",
        "MC contrôle",
        "Largeur IC contrôle",
        "Gain",
        "Temps (s)"
    ]
)

df
```

## Commentaire

La colonne **Gain** mesure l'amélioration apportée par la variable de contrôle :

$$
G=
\frac{\text{largeur IC Monte-Carlo standard}}
{\text{largeur IC variable de contrôle}}.
$$

Lorsque \(G>1\), la méthode à variable de contrôle est plus précise.

On observe généralement une réduction importante de la largeur des intervalles de confiance, ce qui confirme la forte corrélation entre \(f_3\) et \(f_4\).

Le coût supplémentaire reste négligeable puisque les quantités nécessaires au calcul de \(f_4\) sont déjà calculées lors de l'évaluation de \(f_3\).

## Analyse des résultats

On compare les intervalles de confiance obtenus avec et sans réduction de variance.

Pour mesurer quantitativement le gain apporté par la variable de contrôle, on considère le rapport

$$
G
=
\frac{
\text{largeur de l'IC standard}
}{
\text{largeur de l'IC avec réduction de variance}
}.
$$

Lorsque \(G>1\), la méthode à variable de contrôle améliore la précision de l'estimation.

La fonction \(f_4\) étant fortement corrélée à \(f_3\), on observe généralement une diminution sensible de la variance ainsi qu'un rétrécissement des intervalles de confiance.

Le coût supplémentaire reste négligeable puisque le calcul de \(f_4\) utilise les mêmes quantités intermédiaires que celui de \(f_3\).

La méthode de réduction de variance fournit ainsi une estimation plus précise de \(I(f_3)\) pour un temps de calcul pratiquement inchangé.

Dans ce projet, plusieurs méthodes ont été étudiées pour estimer des intégrales de grande dimension.

La méthode de Monte-Carlo standard fournit un estimateur

$$
M_n(f)
=
\frac1n
\sum_{k=1}^{n}

dont la précision est gouvernée par le théorème central limite. Son principal avantage est que son coût ne dépend que linéairement de la dimension.

Nous avons ensuite introduit une méthode de réduction de variance reposant sur une variable de contrôle. Cette approche consiste à exploiter une fonction auxiliaire dont l'intégrale est connue afin de construire un estimateur de variance plus faible.

Les simulations montrent que cette technique permet de réduire significativement la largeur des intervalles de confiance sans augmenter sensiblement le temps de calcul.

Nous avons également étudié le comportement asymptotique de l'intégrale lorsque la dimension tend vers l'infini. La loi forte des grands nombres puis le théorème central limite multidimensionnel permettent d'obtenir une approximation de faible dimension faisant intervenir uniquement un vecteur gaussien bidimensionnel.

Cette approximation conduit à des méthodes numériques beaucoup plus rapides lorsque la dimension devient grande. Les résultats obtenus pour \(d=100\) et \(d=1000\) montrent que l'erreur asymptotique devient rapidement négligeable.

Enfin, nous avons appliqué à nouveau la méthode des variables de contrôle à une intégrale plus complexe. Le choix d'une variable de contrôle fortement corrélée à l'intégrande permet une nouvelle fois de réduire efficacement la variance de l'estimateur.

Ainsi, ce projet met en évidence deux idées fondamentales du calcul Monte-Carlo :

- la réduction de variance permet d'améliorer fortement la précision à coût constant ;
- l'exploitation des résultats asymptotiques peut transformer un problème de très grande dimension en un problème de faible dimension beaucoup plus facile à résoudre.

Ces deux approches constituent des outils essentiels en simulation numérique, en finance quantitative, en statistique computationnelle et plus généralement dans tous les domaines nécessitant le calcul d'intégrales de grande dimension.
